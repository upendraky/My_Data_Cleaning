```python
df.columns = df.columns.str.strip()
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df = df[df["Category"] != "1.9"]
```

```python
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")
```

```python
def convert_size(value):

    value = str(value).strip()

    if value == "Varies with device":
        return np.nan

    if value.endswith("M"):
        return float(value.replace("M", ""))

    elif value.endswith("k"):
        return float(value.replace("k", "")) / 1024

    else:
        return np.nan


df["Size"] = df["Size"].apply(convert_size)
```

```python
df["Installs"] = (
    df["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

df["Installs"] = pd.to_numeric(
    df["Installs"], errors="coerce"
)

df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
)

df["Price"] = pd.to_numeric(
    df["Price"], errors="coerce"
)

df["Type"] = df["Type"].replace({
    "Free": "Free",
    "Paid": "Paid"
})
```

```python
df.loc[~df["Type"].isin(["Free", "Paid"]), "Type"] = np.nan
```

```python
print("Missing Values Before Cleaning:")

print(df.isnull().sum())
```

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].median())
```

```python
df["Reviews"] = df["Reviews"].fillna(0)
```

```python
df["Size"] = df["Size"].fillna(df["Size"].median())
```

```python
df["Installs"] = df["Installs"].fillna(0)
```

```python
df["Price"] = df["Price"].fillna(0)
```

```python
for col in ["Category", "Type", "Content Rating", "Genres"]:
    df[col] = df[col].fillna("Unknown")
```

```python
df.dropna(subset=["App"], inplace=True)
```

```python
df = df[df["Rating"].between(0, 5)]

print("\nMissing Values After Cleaning:")

print(df.isnull().sum())
```

```python
# FEATURE ENGINEERING
df["App_Price_Type"] = np.where(
    df["Price"] == 0, "Free", "Paid"
)
```

```python
df["Last Updated"] = pd.to_datetime(
    df["Last Updated"], errors="coerce"
)
```

```python
df["Update_Year"] = df["Last Updated"].dt.year

# Create rating groups

df["Rating_Group"] = pd.cut(
    df["Rating"],
    bins=[0, 2, 3, 4, 5],
    labels=["Poor", "Average", "Good", "Excellent"],
    include_lowest=True
)

print("Feature Engineering Completed!")
```

```python
category_count = df["Category"].value_counts().head(10)

category_rating = (
    df.groupby("Category")["Rating"]
    .mean()
    .sort_values(ascending=False)
)

category_installs = (
    df.groupby("Category")["Installs"]
    .sum()
    .sort_values(ascending=False)
)

print("Category Analysis Completed!")
```

```python
# VISUALIZATION 
plt.figure(figsize=(12, 6))

sns.barplot(
    x=category_count.values,
    y=category_count.index
)

plt.title("Top 10 Google Play Store Categories")
plt.xlabel("Number of Applications")
plt.ylabel("Category")

plt.tight_layout()
plt.savefig(f"{VISUAL_PATH}/01_top_categories.png")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Rating"],
    bins=20,
    kde=True
)

plt.title("Distribution of App Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Applications")

plt.tight_layout()
plt.savefig(f"{VISUAL_PATH}/02_ratings_distribution.png")
plt.show()


```

```python
plt.figure(figsize=(8, 6))

df["Type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Free vs Paid Applications")
plt.ylabel("")

plt.tight_layout()
plt.savefig(f"{VISUAL_PATH}/03_free_vs_paid.png")
plt.show()
```

```python
plt.figure(figsize=(12, 7))

category_rating.sort_values().plot(
    kind="barh"
)

plt.title("Average Rating by App Category")
plt.xlabel("Average Rating")
plt.ylabel("Category")

plt.tight_layout()
plt.savefig(f"{VISUAL_PATH}/04_category_ratings.png")
plt.show()
```

```python
top_apps = (
    df.sort_values("Installs", ascending=False)
    .drop_duplicates(subset="App")
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_apps,
    x="Installs",
    y="App"
)

plt.title("Top 10 Most Installed Apps")
plt.xlabel("Number of Installs")
plt.ylabel("Application")

plt.tight_layout()
plt.savefig(f"{VISUAL_PATH}/05_top_installed_apps.png")
plt.show()


```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Size"],
    bins=30,
    kde=True
)

plt.title("Distribution of App Sizes")
plt.xlabel("App Size (MB)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig(f"{VISUAL_PATH}/07_app_size_distribution.png")
plt.show()
```
